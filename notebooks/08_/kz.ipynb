```python
import csv
```

```python
with open("../../data/processed/unblocked_pairs_2.csv") as f:
    reader = csv.reader(f)
    results = [tuple(map(int, row)) for row in reader]
```

```python
results_with_reverse = results + [(y, x) for (x, y) in results]
```

```python
from collections import Counter

x_counts = Counter(x for x, y in results_with_reverse)
print(x_counts)
```

```python
len(x_counts.values())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(x_counts.values(), bins=30, edgecolor='black')
plt.xlabel('Count')
plt.ylabel('Frequency')
plt.title('Histogram of x_counts')
plt.show()
```
